```python
SEED_NUM = 7 
STUDENT_MAJOR = "BSECE"
FAVORITE_ARTIST = "Paramore" 
LOCAL_HAZARD = "FLOOD"
CONTROL_NUM = max(3, SEED_NUM) 
```

```python
from collections import defaultdict, Counter

stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

custom_entries = [("TUP Manila", "BSECE") for _ in range(10)]
stream.extend(custom_entries)

campus_map = defaultdict(list)
for campus, program in stream:
    campus_map[campus].append(program)

manila_counts = Counter(campus_map["TUP Manila"])
top_program, frequency = manila_counts.most_common(1)[0]

print(f"CONTROL_NUM Used: {CONTROL_NUM}")
print(f"STUDENT_MAJOR Used: {STUDENT_MAJOR}")
print(f"Expanded Stream Length: {len(stream)}")
print(f"Total Applications Processed for TUP Manila: {len(campus_map['TUP Manila'])}")
print(f"Top Requested Program at TUP Manila: {top_program}")
print(f"Exact Frequency of Top Program: {frequency}")
```

```python
festival = {"Ben&Ben", "SB19", "Bini", "Eraserheads", FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"}

user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST, f"Indie Artist {CONTROL_NUM}"}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

new_artists = {"Mayday Parade", "The Maine", "Neck Deep", "State Champs", "Real Friends", "Knuckle Puck", "Hot Mulligan"}
festival.update(new_artists)

intersection = user_a.intersection(festival)
union = user_a.union(festival)
jaccard_a = (len(intersection) / len(union)) * 100

intersection_b = user_b.intersection(festival)
union_b = user_b.union(festival)
jaccard_b = (len(intersection_b) / len(union_b)) * 100

dealbreakers = user_a.difference(festival)

print(f"Total Artists in Expanded Festival: {len(festival)}")
print(f"Length of Intersection (User A & Festival): {len(intersection)}")
print(f"Length of Union (User A & Festival): {len(union)}")
print(f"User A Jaccard Similarity (%): {jaccard_a:.2f}%")
print(f"User B Jaccard Similarity (%): {jaccard_b:.2f}%")
print(f"User A Dealbreaker Artists: {dealbreakers}")
```

```python
from collections import deque, Counter

buffer = deque(maxlen=5)

burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, "LOCAL_HAZARD"),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

last_ts = CONTROL_NUM + 6
for i in range(1, 10):
    burst.append((last_ts + i, "WEATHER" if i % 2 == 0 else LOCAL_HAZARD))

weather_appended = "N"
hazard_appended = "N"

for ts, category in burst:
    current_categories = [item[1] for item in buffer]
    counts = Counter(current_categories)
    
    if counts[category] < 2:
        buffer.append((ts, category))
        if ts == (CONTROL_NUM + 3) and category == "WEATHER": weather_appended = "Y"
        if category == LOCAL_HAZARD: hazard_appended = "Y"

final_sum = sum(item[0] for item in buffer)

print(f"LOCAL_HAZARD Used: {LOCAL_HAZARD}")
print(f"Total Burst Size: {len(burst)}")
print(f"Did 'WEATHER' at Timestamp {CONTROL_NUM + 3} append? (Y/N): {weather_appended}")
print(f"Did {LOCAL_HAZARD} append? (Y/N): {hazard_appended}")
print(f"Final Deque Length: {len(buffer)}")
print(f"Final Sum of Timestamps in Deque: {final_sum}")
print(f"Exact Final Deque Output: {list(buffer)}")
```
